# RAGAS Evaluation Dataset — E-Bike X1

This notebook creates a small **RAG evaluation dataset** using `SingleTurnSample` and `EvaluationDataset` from RAGAS.

Each sample contains:

- `user_input` — the user's question
- `retrieved_contexts` — context retrieved by the RAG system
- `response` — answer generated by the application
- `reference` — expected/ground-truth answer

> This is a local evaluation-dataset exercise. It does not call AWS and does not require AWS credentials or an IAM role.


## 1. Install RAGAS

```python
# Run this cell if RAGAS is not already installed.
%pip install -q ragas
```

## 2. Import RAGAS dataset classes

```python
from ragas import SingleTurnSample, EvaluationDataset
```

## 3. Create the first evaluation sample

```python
sample1 = SingleTurnSample(
    user_input="What is the battery warranty of the E-Bike X1?",
    retrieved_contexts=[
        "The E-Bike X1 battery includes a 2-year manufacturer warranty."
    ],
    response="The E-Bike X1 battery has a 2-year warranty.",
    reference="The battery warranty is 2 years."
)

sample1
```

### What this sample represents

The RAG application retrieves information about the battery warranty and generates an answer from that context.

A good evaluation should check whether:

1. The retrieved context is relevant to the question.
2. The generated response is supported by the retrieved context.
3. The response correctly answers the user's question.


## 4. Create the second evaluation sample

```python
sample2 = SingleTurnSample(
    user_input="How long does the E-Bike X1 take to charge?",
    retrieved_contexts=[
        "A full charge of the E-Bike X1 takes approximately 4 hours."
    ],
    response="The E-Bike X fu1 takes about 4 hours tolly charge.",
    reference="Approximately 4 hours."
)

sample2
```

## 5. Build the evaluation dataset

```python
dataset = EvaluationDataset(samples=[sample1, sample2])

dataset
```

## 6. Inspect the samples

```python
for i, sample in enumerate(dataset.samples, start=1):
    print(f"--- Sample {i} ---")
    print("Question :", sample.user_input)
    print("Context  :", sample.retrieved_contexts)
    print("Response :", sample.response)
    print("Reference:", sample.reference)
    print()
```

## Teaching takeaway

The `EvaluationDataset` is the test dataset that can later be passed to RAGAS evaluation metrics.

For each test case, we keep the **question, retrieved evidence, generated answer, and expected answer** together. This makes it possible to evaluate different parts of a RAG pipeline, such as retrieval quality and answer quality.
